## Section 1: Image Conversion and Preprocessing
This notebook contains the Python implementation for converting 112 hand-drawn .pgm images into 45x45 binary CSV matrices. 

**Key Process:**
* **Input:** PGM ASCII files (45x45) drawn in GIMP
* **Threshold:** 128 (Values < 128 become 0; others become 1)
* **Output:** CSV files named `40457802_[label]_[index].csv`

```python
import os
import numpy as np

# Configuration for CSV file name format
STUDENT_NUM = "40457802"
DIR = "./images"

# Converting the PGM files to CSV
def pgm_to_csv(file_path, output_path):
    with open(file_path, 'r') as file:

        lines = file.readlines()
        
        # Extracting the pixel data from the PGM
        # We skip the first three lines and start from the fourth for the pixel data
        pixels = []
        for line in lines[4:]:
            pixels.extend(line.split())
        
        # Converting strings to integers for CSV
        pixels = [int(p) for p in pixels]
        
        # Creating 45x45 matrix
        matrix = np.array(pixels).reshape(45, 45)
        

        # Converting any value less than 128 to 0, and greater than or equal to 128 to 1
        # This creates a matrix of 1s and 0s representing the image
        binary_matrix = np.where(matrix < 128, 0, 1)
        
        # Saving as CSV (Comma delimited, removing decimals)
        np.savetxt(output_path, binary_matrix, delimiter=",", fmt='%d')


# Processing all files
for file_name in os.listdir(DIR):
    if file_name.endswith(".pgm"):
        input_path = os.path.join(DIR, file_name)
        
        # Ensuring the correct output format: STUDENTN_LABEL_INDEX.csv
        # PGM files already saved in this format: LABEL_INDEX.pgm
        csv_filename = file_name.replace(".pgm", ".csv")
        csv_filename = f"{STUDENT_NUM}_{csv_filename}"
            
        output_path = os.path.join(DIR, csv_filename)
        
        pgm_to_csv(input_path, output_path)
        print(f"Processed: {csv_filename}")

print("\nTask 1.2 Complete. All images converted to CSV matrices.")
```
